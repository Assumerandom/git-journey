```python
import random

def guessing_game():
    print("Hello, pick a number 1-10")
    rnd_num = random.randint(1, 10)
    
    while True:
        guess = int(input("Your guess: "))
        if guess == rnd_num:
            print("oh wow, you got number :D")
            break
        else: 
            print("Oof, wrong! Try again D:")

guessing_game()
```
